```python
# Dati (le stesse colonne B e C del foglio, ora 0-23)
produzione = [0,0,0,0,0,0,0,30,100,200,300,380,400,380,320,240,140,60,10,0,0,0,0,0]
prezzo = [80,80,80,80,80,80,80,50,50,50,50,50,50,50,50,50,50,50,120,120,120,120,80,80]

# Parametri (le celle N1:N4 del foglio)
efficienza = 0.9
soglia_bassa = 60
soglia_alta = 100
capacita = 400
potenza = 100

ricavo_senza = 0
ricavo_con = 0
caricato = 0
scaricato = 0

for i in range(24):  # una volta per ogni ora, come ogni riga del foglio
    ricavo_senza += produzione[i] * prezzo[i] / 1000

    carica = 0
    scarica = 0
    # carico se il prezzo è basso e c'è abbastanza produzione
    if prezzo[i] < soglia_bassa and produzione[i] >= 100:
        carica = min(potenza, capacita - caricato)
        caricato += carica
    # scarico se il prezzo è alto
    if prezzo[i] >= soglia_alta:
        scarica = min(potenza, caricato - scaricato)
        scaricato += scarica

    ricavo_con += (produzione[i] - carica + scarica * efficienza) * prezzo[i] / 1000

print("Senza batteria:", round(ricavo_senza, 1))
print("Con batteria:", round(ricavo_con, 1))
print("Guadagno:", round(ricavo_con - ricavo_senza, 1))
```
